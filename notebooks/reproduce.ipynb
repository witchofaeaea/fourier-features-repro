## Plot: test PSNR vs σ

```
import csv, matplotlib.pyplot as plt
rows=[r for r in csv.DictReader(open('results/natural/summary.csv')) if r['mapping']=='gaussian']
rows.sort(key=lambda r: float(r['sigma']))
s=[float(r['sigma']) for r in rows]; m=[float(r['test_psnr_mean']) for r in rows]; sd=[float(r['test_psnr_std']) for r in rows]
plt.plot(s,m,'o-'); plt.fill_between(s,[a-b for a,b in zip(m,sd)],[a+b for a,b in zip(m,sd)],alpha=.3)
plt.xscale('log'); plt.xlabel('σ'); plt.ylabel('test PSNR (dB)'); plt.grid(alpha=.3); plt.show()
```
